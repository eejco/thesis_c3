```python
import gsw
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import seaborn as sns
import xesmf as xe
import glob

fname_zhou = '../data/CLIM_3D_forJenny.nc'
fname_spira = '../data/qc_profile_ds_2dbar.nc'
fname_glorys1 = '../data/GLORYS/glorys_001.nc'
fname_glorys2 = '../data/GLORYS/glorys_002.nc'
fname_sose = '../data/SOSE/adelie3d/Jenny_AdelieSlices'
```

```python
extent = [138, 150,-68,-64]
extent2 = [139,142,-67,-66]
```

```python
ds = xr.open_dataset('profiles_full.nc')

ds_shelf = ds.where(ds.elevation > -1000,drop=True)
```

```python
t1 = ds_shelf.time.min()
t2 = ds_shelf.time.max()
```

```python
#get sose
sose=xr.open_mfdataset(glob.glob(fname_sose + '/*.nc'))
```

```python
t0 = sose.time.min()
```

```python
# prepare sose
# crop to region
lats = sose.YC.load()
lons = sose.XC.load()
depth = sose.Z.load()
mask = (lats >= extent2[2]) & (lats <= extent2[3]) & (lons >= extent2[0]) & (lons <= extent2[1]) & (depth < 1000)
sose_shelf = sose.where(mask,drop=True)

# loiads of zeros
idx=sose_shelf.SALT.load() > 5
sose_shelf = sose_shelf.where(idx,np.nan)

# make monthly
sose_shelf=sose_shelf.resample({'time':'MS'}).mean().sel(time=slice(t1,t2))
sose_shelf['pres'] = gsw.conversions.p_from_z(sose_shelf.Z,sose_shelf.YC.mean())

sose_shelf_forinterp = sose_shelf.rename({
    "YC": "lat",
    "XC": "lon"
}).swap_dims({'Z':'pres'}).drop_vars('Z')

sose_shelf_forinterp.to_netcdf('sose_shelf_forinterp.nc')
```

```python
# glorys first. need to interpolate before loading out of dask
ds1 = xr.open_dataset(fname_glorys1)
ds2 = xr.open_dataset(fname_glorys2)
glorys = xr.merge([ds1,ds2]).sel(time=slice(t1,t2))
glorys_shelf = glorys.sel(latitude=slice(extent2[2],extent2[3])).sel(longitude=slice(extent2[0],extent2[1])).where(glorys.depth<1000,drop=True)

glorys_shelf['pres'] = gsw.conversions.p_from_z(-glorys_shelf.depth,glorys_shelf.latitude.mean())

glorys_shelf_forinterp = glorys_shelf.rename({
    "latitude": "lat",
    "longitude": "lon"
}).swap_dims({'depth':'pres'}).drop('depth')

glorys_shelf_forinterp.to_netcdf('glorys_shelf_forinterp.nc')
```

```python
# load nemo data
catalog = OceanDataCatalog(catalog_name="noc-model-stac")
nemo = catalog.open_dataset(id='noc-npd-era5/npd-eorca12-era5v1/gn/T1m_4d',variable_names=['thetao_con','so_abs'])

# prepare nemo
# crop to region
lats = nemo.nav_lat.load()
lons = nemo.nav_lon.load()
depth = nemo.deptht.load()
mask = (lats >= extent2[2]) & (lats <= extent2[3]) & (lons >= extent2[0]) & (lons <= extent2[1]) & (depth < 1000)
nemo_shelf = nemo.where(mask,drop=True)
# make monthly
nemo_shelf=nemo_shelf.resample({'time_counter':'MS'}).mean().sel(time_counter=slice(t1,t2))
nemo_shelf['pres'] = gsw.conversions.p_from_z(-nemo_shelf.deptht,nemo_shelf.nav_lat.mean())

nemo_shelf_forinterp = nemo_shelf.rename({
    "time_counter":"time",
    "nav_lat": "lat",
    "nav_lon": "lon"
}).swap_dims({'deptht':'pres'}).drop('deptht')

nemo_shelf_forinterp.to_netcdf('nemo_shelf_forinterp.nc')
```

```python
glorys_shelf_forinterp = xr.open_dataset('glorys_shelf_forinterp.nc')
nemo_shelf_forinterp = xr.open_dataset('nemo_shelf_forinterp.nc')
sose_shelf_forinterp = xr.open_dataset('sose_shelf_forinterp.nc')
```

```python
theta_interp = sose_shelf_forinterp.THETA.interp(time=ds_shelf.time.data,
                                                     pres=ds_shelf.pres.data,
                                                     lat=("time",ds_shelf.lat.data),
                                                     lon=("time",ds_shelf.lon.data),    
                                                     method='linear')

salt_interp = sose_shelf_forinterp.SALT.interp(time=ds_shelf.time.data,
                                             pres=ds_shelf.pres.data,
                                             lat=("time",ds_shelf.lat.data),
                                             lon=("time",ds_shelf.lon.data),
                                             method='linear')

sose_asal = gsw.conversions.SA_from_SP(salt_interp,salt_interp.pres,salt_interp.lon,salt_interp.lat)
sose_ctemp = gsw.conversions.CT_from_pt(sose_asal,theta_interp)
```

```python
# chunk for speed
glorys_shelf_forinterp = glorys_shelf_forinterp.chunk({
    "time": 10,
    "lat": 50,
    "lon": 50,
    "pres": 42   # full depth is fine
})

# chunk for speed
nemo_shelf_rename = nemo_shelf_forinterp.chunk({
    "time": 10,
    "y": 50,
    "x": 50,
    "pres": 42   # full depth is fine
})

ds_shelf = ds_shelf.chunk({"time": 200})
```

```python
thetao_interp = glorys_shelf_forinterp.thetao.interp(time=ds_shelf.time.data,
                                                     pres=ds_shelf.pres.data,
                                                     lat=("time",ds_shelf.lat.data),
                                                     lon=("time",ds_shelf.lon.data),
                                                     method='linear')


so_interp = glorys_shelf_forinterp.so.interp(time=ds_shelf.time.data,
                                             pres=ds_shelf.pres.data,
                                             lat=("time",ds_shelf.lat.data),
                                             lon=("time",ds_shelf.lon.data),
                                             method='linear')

glorys_asal = gsw.conversions.SA_from_SP(so_interp,so_interp.pres,so_interp.lon,so_interp.lat)
glorys_ctemp = gsw.conversions.CT_from_pt(so_interp,thetao_interp)
```

```python
# nemo is curvilinear so do time first
nemo_shelf_tinterp = nemo_shelf_rename.interp(time=ds_shelf.time, method="linear")

# then lat and lon
ds_in = xr.Dataset({
    "lat": (("y", "x"), nemo_shelf_tinterp.lat.data),
    "lon": (("y", "x"), nemo_shelf_tinterp.lon.data),
})

ds_out = xr.Dataset({
    "lat": ("points", ds_shelf.lat.data),
    "lon": ("points", ds_shelf.lon.data)
})

regridder = xe.Regridder(ds_in, ds_out, "bilinear",locstream_out=True)

nemo_regridded = regridder(nemo_shelf_tinterp)

# THIS INTRODUCES LOTS OF ZEROS!
```

```python
idx=nemo_regridded.so_abs.load() > 5
nemo_regridded = nemo_regridded.where(idx,np.nan)

# rejig lat and lon and interp depth
nemo_interp = (
    nemo_regridded
    .assign_coords({
        "time": ds_shelf.time,
        "lat": ("time", ds_shelf.lat.data),
        "lon": ("time", ds_shelf.lon.data),
    })
)
nemo_interp2=nemo_interp.isel(points=xr.DataArray(
    np.arange(len(nemo_interp.time)),
    dims="time"
)).interp(pres=ds_shelf.pres,method='linear')
```

```python
profiles_shelf_ts = ds_shelf.expand_dims({'source':['Observations']}).drop_vars(['mld','temp','psal','rho','elevation','month','year','sigma0'])
nemo_shelf_ts = nemo_interp2.rename({'thetao_con':'ctemp','so_abs':'asal'}).expand_dims({'source':['NEMO']})
glorys_shelf_ts = xr.Dataset({'ctemp':glorys_ctemp, 'asal': glorys_asal}).expand_dims({'source':['GLORYS']})
sose_shelf_ts = xr.Dataset({'ctemp':sose_ctemp, 'asal': sose_asal}).expand_dims({'source':['SOSE']}).drop_vars(['rA','Depth','drF','hFacC'])
```

```python
pgn = xr.concat([profiles_shelf_ts,glorys_shelf_ts,nemo_shelf_ts,sose_shelf_ts],dim='source')
```
