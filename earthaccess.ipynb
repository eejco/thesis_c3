```python
import earthaccess
import numpy as np
earthaccess.login()
```

```python
results = earthaccess.search_data(
    short_name='ATL06',
    bounding_box=(-10, 20, 10, 50),
    temporal=("1999-02", "2019-03"),
    count=10
)
```

```python
results = earthaccess.search_data(
    short_name='NSIDC-0051',
    bounding_box=(140,-70,150,-60),
    temporal=("2002-01", "2022-01")
)
```

```python
files = [g['meta']['native-id'] for g in results]
is_monthly = [(len(f)<42) for f in files]
results_monthly = np.array(results)[is_monthly]
filtered_results=[r for r, k in zip(results,is_monthly) if k]
```

```python
files = earthaccess.download(filtered_results, "../data/seaicetest")
```
